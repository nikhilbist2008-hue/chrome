```python
 import math

 x=int(input("enter a number"))
 expression=(
    math.exp(x)
    +math.sin(x)
    +math.log2(x**2)
    +math.log(math.exp(x))
    +math.log10(x)
    +math.pi*(x**3)
    +math.atan(x**3)
    +math.cos(x)
    +math.tau
    +(math.exp(x)/math.exp(-x+1))
    +math.comb(5,2)
    +math.perm(6,3)
    +math.factorial(10)
 )
 print(f"the following expression is evaluated to{round(expression,2)}")
 print(f"the following expression is evaluated to{expression:.2f}")
```
